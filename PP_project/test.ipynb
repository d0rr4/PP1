```python

# --- DIAGNOSTIC CELL: Check for ID Mismatch ---
# This cell will compare the protein identifiers from your HDF5 embedding file
# with the identifiers in your CSV annotation file. A low overlap is the most
# likely cause of missing annotations in the web UI.

import h5py
import pandas as pd

# --- Configuration ---
H5_FILE = "embeddings/cath_s40_prott5_embeddings.h5"
CSV_FILE = "datasets/cath_annotation.csv"
CSV_ID_COLUMN = "Entry"  # The column in your CSV that contains the protein IDs

# --- Read IDs from HDF5 File ---
with h5py.File(H5_FILE, "r") as f:
    h5_ids = set(f.keys())

# --- Read IDs from CSV File ---
csv_df = pd.read_csv(CSV_FILE)
csv_ids = set(csv_df[CSV_ID_COLUMN])

# --- Report ---
print(f"--- Comparing IDs from '{H5_FILE}' and '{CSV_FILE}' ---")
print(f"Found {len(h5_ids)} protein IDs in the HDF5 embedding file.")
print(f"Found {len(csv_ids)} protein IDs in the CSV annotation file (from column '{CSV_ID_COLUMN}').")
print("-" * 60)
print(f"Overlap: There are {len(h5_ids.intersection(csv_ids))} matching IDs between the two files.")
print("-" * 60)

# Show a sample of IDs from each file to help spot formatting differences
print("Sample IDs from HDF5 file:", list(h5_ids)[:5])
print("Sample IDs from CSV file: ", list(csv_ids)[:5])

if not h5_ids.intersection(csv_ids):
    print("\\n\\n*** WARNING: Zero overlapping IDs found. ***")
    print("This is the reason your annotations are not appearing.")
    print("The identifiers in your H5 file and your CSV file do not match.")
    print("Please check the ID formats. For example, one file might have 'P12345' while the other has 'sp|P12345|3FTX_NAJNA'.")
```

```python

# Step 1: Inspect the generated bundle to confirm missing settings
# We will see that the 'cath' bundle has 3 parts (no settings), while the 'toxins' one has 4.

import os
from pathlib import Path
import json

def inspect_bundle(bundle_path):
    if not os.path.exists(bundle_path):
        print(f"Bundle not found at: {bundle_path}")
        return
    with open(bundle_path, "rb") as f:
        content = f.read()
    parts = content.split(b"---PARQUET_DELIMITER---")
    print(f"Inspecting '{bundle_path}': Found {len(parts)} parts.")
    if len(parts) == 4 and parts[3]:
        print("Bundle contains a settings part.")
    else:
        print("Bundle does NOT contain a settings part (this is why colors are missing).")

# Path to your generated bundles
cath_bundle_path = "protspace_results/cath/data.parquetbundle"
toxins_bundle_path = "protspace_results/toxins/data.parquetbundle"

inspect_bundle(cath_bundle_path)
inspect_bundle(toxins_bundle_path)

# Step 2: Generate and apply color settings to the CATH bundle
# We'll use the `protspace style` command to do this.

# First, generate a template. This creates a JSON structure with all your
# CSV columns and their most frequent values, ready for styling.
print("\\n--- Generating style template for CATH bundle ---")
style_template_json = !protspace style --generate-template {cath_bundle_path}
style_template = json.loads("".join(style_template_json))

# --- NEW: Explicitly define a color palette ---
# Instead of just using the template, let's add a specific color palette
# to the 'cath_class' and 'architecture' annotations.
# This gives the command more explicit instructions.
print("\\n--- Adding explicit color palettes ---")
palettes = {
    "categorical": [
        "#a6cee3", "#1f78b4", "#b2df8a", "#33a02c", "#fb9a99", "#e31a1c",
        "#fdbf6f", "#ff7f00", "#cab2d6", "#6a3d9a", "#ffff99", "#b15928"
    ]
}
if 'cath_class' in style_template:
    style_template['cath_class']['palette'] = palettes
    print("Added palette to 'cath_class'")

if 'architecture' in style_template:
    style_template['architecture']['palette'] = palettes
    print("Added palette to 'architecture'")

# Now, apply these more detailed styles back to the bundle.
# This creates a *new* bundle file with the settings included.
output_styled_bundle_path = "protspace_results/cath/data_styled_explicit.parquetbundle"
!protspace style {cath_bundle_path} {output_styled_bundle_path} --annotation-styles '{json.dumps(style_template)}'

print(f"\\nSuccessfully created styled bundle at: {output_styled_bundle_path}")

# Step 3: Verify the new bundle
print("\\n--- Verifying the new styled bundle ---")
inspect_bundle(output_styled_bundle_path)

print(f"\\nUpload the NEW file '{output_styled_bundle_path}' to protspace.app. This one has explicit color palettes defined.")
```

```python
import io
import pandas as pd
import pyarrow.parquet as pq
from pathlib import Path

# --- Config (Matching your setup) ---
DATASET     = "toxins"
RESULTS_DIR = Path("protspace_results") / DATASET
BUNDLE      = RESULTS_DIR / "data.parquetbundle"
DELIMITER   = b"---PARQUET_DELIMITER---"
LABEL_COL   = "protein_families"

print(f"Opening bundle for dataset: {DATASET}...")

# 1. Read and unpack the bundle
with open(BUNDLE, "rb") as f:
    parts = f.read().split(DELIMITER)

# 2. Parse the metadata dataframe
annot_df = pq.read_table(io.BytesIO(parts[0])).to_pandas()

# 3. Inspect raw labels
if LABEL_COL not in annot_df.columns:
    print(f"Error: Column '{LABEL_COL}' not found! Available columns are: {list(annot_df.columns)}")
else:
    # Get all unique valid string labels
    raw_labels = annot_df[LABEL_COL].dropna().astype(str).unique()
    
    # Filter for the ones that actually contain a pipe '|'
    piped_labels = [label for label in raw_labels if "|" in label]
    
    print(f"\nTotal unique labels in dataset : {len(raw_labels)}")
    print(f"Labels containing '|'          : {len(piped_labels)}")
    print("=" * 70)
    
    if piped_labels:
        print(f"Showing up to 20 examples of modified strings:\n")
        for i, raw in enumerate(piped_labels[:20], 1):
            # Mirror the exact logic from your main script
            modified = raw.split("|")[0].strip()
            
            print(f"Example {i}:")
            print(f"  [RAW INPUT] -> \"{raw}\"")
            print(f"  [MODIFIED]  -> \"{modified}\"")
            print("-" * 70)
    else:
        print("No labels containing '|' were found in this specific dataset bundle.")
        print("\nHere is a random sample of the raw labels that DO exist:")
        for raw in raw_labels[:10]:
            print(f"  -> \"{raw}\"")
```
