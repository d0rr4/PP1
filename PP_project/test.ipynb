```python
import io
import json
import pyarrow.parquet as pq
from pathlib import Path

DELIMITER = b"---PARQUET_DELIMITER---"

TABLE_NAMES = [
    "selected_annotations",
    "projections_metadata",
    "projections_data",
]

path = Path("/mnt/c/Users/janni/OneDrive/repos/PP1/data/3FTx/parquetbundle/3ftx_with_db_styled.parquetbundle")

# read file
with open(path, "rb") as f:
    content = f.read()

# split bundle
parts = content.split(DELIMITER)

print("Number of bundle parts:", len(parts))
print()

# inspect core tables
for name, blob in zip(TABLE_NAMES, parts[:3]):
    print("=" * 60)
    print(name)

    table = pq.read_table(io.BytesIO(blob))

    print("rows:", table.num_rows)
    print("cols:", table.num_columns)
    print("schema:")
    print(table.schema)

    print("\npreview:")
    print(table.slice(0, 5).to_pandas())

    print()

# inspect settings if present
if len(parts) == 4 and parts[3]:
    print("=" * 60)
    print("settings")

    table = pq.read_table(io.BytesIO(parts[3]))
    settings_json = table.column("settings_json")[0].as_py()
    settings = json.loads(settings_json)

    print("top-level keys:", list(settings.keys()))
    print()
    print(json.dumps(settings, indent=2)[:2000])  # truncate if huge
else:
    print("No settings part found")
```
